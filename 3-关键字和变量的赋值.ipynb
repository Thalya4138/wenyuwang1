***<font color=FF0000>赋值语句</font>***

**<font color=#0080FF>赋值语句的 3 种基本形式：</font>**  
1. 变量 = 表达式
2. 变量1, 变量2,... = 表达式1, 表达式2，...  (被称为平行赋值、并行赋值)
3. 变量1 = 变量2 = 表达式            (被称为串行赋值、合并赋值)

```python
num = 123456    ## 执行赋值语句，不会产生计算结果，是在建立变量名与对象之间的约束关系
```

```python
print(num)
```

```python
i, j = 3, 5.5   ## 平行赋值
print(i, j)
```

```python
i, j = j, i     ## 利用平行赋值，可以很方便地交换变量值
print(i, j)      
```

```python
i = j = k = 10  ## 串行赋值
print(i, j, k)
```

**<font color=#0080FF>实例</font>：已知三角形边长为 3, 4, 5，求三角形的面积**

```python
from math import sqrt
s = (3 + 4 + 5) / 2             ## 引入变量 s 记录/保存半周长的值
sqrt(s * (s - 3) * (s - 4) * (s - 5))
```
