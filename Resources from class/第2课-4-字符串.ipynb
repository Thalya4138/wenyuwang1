**<font color=#0080FF>raw string</font>**  
- 在一个字符串字面量的前面，加 <font color=yellow>前缀 r 或 R</font>，将使得该字符串里的转义序列失效
- 使用场景：字符串的部分内容正好为某个转义序列；例如，表示文件路径的字符串经常出现这种情况    

```python
print('c:\tempfile')  ## 其中的 \t 会作为一个转义序列，表示制表符
```

```python
print(r'c:\tempfile') ## 其中的 \t 不会被作为转义序列，而是两个字符构成的序列
```

```python
print('c:\\tempfile') 
## 和上一个字符串字面量等价，其中的 \\ 是一个转义序列，表示反斜符
```

**<font color=#0080FF>f-string —— formatted string literals 格式化字符串字面量</font>**

+ https://docs.python.org/zh-cn/3.8/reference/lexical_analysis.html#formatted-string-literals
+ https://realpython.com/python-f-strings/
+ https://mlln.cn/2018/05/19/python3%20f-string%E6%A0%BC%E5%BC%8F%E5%8C%96%E5%AD%97%E7%AC%A6%E4%B8%B2%E7%9A%84%E9%AB%98%E7%BA%A7%E7%94%A8%E6%B3%95/

+ F-strings provide a way to embed expressions inside string literals, using a minimal syntax. It should be noted that an f-string is really an expression evaluated at run time, not a constant value. In Python source code, an f-string is a literal string, prefixed with ‘f’, which contains expressions inside braces. The expressions are replaced with their values. 

<b><font color=orange>关于 f-string:</font></b>
+ 花括号外面的部分，按普通字符串字面量处理 (其中可以出现转义序列)
+ 花括号里面的表达式被称为 “替换字段”，左右花括号是替换字段的标志
    - 在花括号里，可以出现任何有效的 Python 表达式，例如，变量名、算术表达式、函数调用表达式等
+ 执行时，系统会把花括号中的表达式作为普通表达式进行求值，再把结果按照所要求格式生成替代字符

```python
print(f'{3 + 4}')
```

```python
from math import pi, e
print(f'{pi}, {e}')
```

<p></p>

<b><font color=orange>f-string 中，替换字段 {表达式 = } 可以实现同时显示表达式文本，'=' 及其求值结果值</font></b>
+ 适用于 log 法调试程序

```python
print(f'{3 + 4 = }')   # 等价于 print(f'3 + 4 = {3 + 4}')

print(f'{pi = }, {e = }')
```

<p></p>

<b><font color=orange>f-string 的替换字段中，可以使用各种控制输出格式的字符:</font></b>
+ :d  :b  :o  :x     以 十进制 / 二进制 / 八进制 /十六进制 输出整数
+ :<  :>  :^         输出时，居左 / 右 / 中对齐
+ :e  :E             以科学计数法输出
+ :f  :F             输出小数
+ :#.#               设置输出最小宽度和小数点位数
+ :,  :_             设置千分位分隔符

```python
n = 1250
print(f'There are {n} students.')
print(f'There are {n:b} students.')
print(f'There are |{n:^10}| students.')
print(f'There are {n:,} students.')
```

```python
print(f'{pi = :8.4f}')
print(f'{pi = :08.4f}')
print(f'{pi = :<8.4f} ')
```
