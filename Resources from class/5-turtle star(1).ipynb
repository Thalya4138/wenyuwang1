***<font color=FF0000>海龟绘图 turtle</font>***

- 模拟海龟在沙滩上爬行所描绘的轨迹，从 LOGO 语言借鉴而来
- Ref：https://docs.python.org/zh-cn/3/library/turtle.html
- demo：IDLE Help 或 https://docs.python.org/zh-cn/3/library/turtle.html#module-turtledemo

<p></p>

**<font color=#0080FF>简单绘图程序模板</font>** 
1. 导入 turtle 模块
2. 新建一只海龟
   - 可做一些初始化设定
3. 用作图语句绘图
4. 结束作图
   - 可选隐藏海龟 

```python
# 1. 导入模块
import turtle  

# 2. 新建一只海龟，并做一些设定
t = turtle.Turtle() 
t.color('red')         # 画笔颜色
t.fillcolor('yellow')  # 填充颜色
t.speed(8)             # 海龟移动速度

# 3. 开始绘图
t.begin_fill()         # 在绘制需要填充颜色的形状之前调用

while True:
	t.forward(200)     # 海龟前进 100 像素，画出一条线段
	t.left(170)        # 向左转 170 度 (逆时针)
	if abs(t.pos()) < 1:  # t.pos() 返回海龟当前的坐标 
		break

t.end_fill()           # 填充绘制好的形状

# 4. 结束绘图
t.hideturtle()         # 隐藏海龟
turtle.done()          # 一个海龟绘图程序的结束语句
```

<p></p>

**<font color=#0080FF>一些说明：</font>**

- 初始，海龟于画布平面的 (0, 0) 点，方向朝东/右
- 基本绘制和控制函数/方法：
    + pensize(): 设置画笔粗细
    + penup(), pendown(): 画笔抬起、落下
    + speed(): 设置画笔速度
    + color()：画笔颜色
    + forward(n) | fd(n), backward(n) | bk(n)：前进/后退 n 个像素
    + right(n) | rt(n), left(n) | lt(n): 右转/左转 n 个单位 (默认单位为角度)
    + pos(): 得到海龟在屏幕中的位置，初始位置在 (0, 0)
    + goto(x, y): 将海龟直接定位到一个绝对坐标
