```python
type(100)
```

**<font color=#0080FF> int(x=0) </font>**  
+ Return an integer object constructed from a number or string x, or return 0 if no arguments are given
+ If x is a floating point numbers, this truncates towards zero (即趋零取整) 

```python
int(4.56)
```

```python
int(-2.7)
```

```python
float(123)
```

<p></p>

__<font color = #0080FF> / --> floating division </font>__  
__<font color = #0080FF> // -> floor division、integer division、整除、地板除 </font>__  

+ The resultant value is a whole integer, though the result’s type is not necessarily int. 
+ // cannot be used for complex numbers.
    

```python
10 / 3
```

```python
10 // 3
```

```python
10.0 / 3.0
```

```python
10.0 // 3.0
```

<p></p>

__<font color=#0080FF> x % y --> remainder of x / y </font>__  

+ 结果的正负，总是与第二个运算对象一致（或是为零）
+ 结果的绝对值，一定小于第二个运算对象的绝对值 
+ 运算对象不能是复数

**<font color=yellow>解释器保证：被除数 = 除数 * 整除商 + 余数</font>**

```python
10 % 3
```

```python
10 % -3
```

```python
-10 % 3
```

```python
-10 % -3
```

<p></p>

**<font color=#0080FF>一个合法的表达式描述了一个计算过程；Python 解释器对一个表达式求值，则得到一个值</font>**  
  
**<font color=#0080FF>写算术表达式时，Python 提倡：</font>**  
+ 在一般运算符的前后，各加一个空格
+ 在正负号与其后的运算对象之间，不留空格 
+ 乘幂运算符的前后都不留空格

```python
3 + +5
```

```python
3 +* 5
```

<p></p>

**<font color=#0080FF>NOTE</font>：对于合法表达式，在计算时也可能出错！**

```python
12 / 0
```

**<font color=#0080FF>NOTE</font>：Python 中的整数计算是“精确的”，总得到精确的整数结果 ('真'除法 / 除外)**

```python
2**1024
```

```python
10 / 3
```

<p></p>

**<font color=#0080FF>Python 的内置浮点数采用 IEEE 754 双精度浮点数标准</font>**  
+ 具有 16-17 位十进制精度
+ 表示范围约为 ±($5×10^{-324}$ — $1.7×10^{308}$)
    
**<font color=#0080FF>浮点数的计算是近似计算，得到有限精度的近似结果</font>**
+ 绝对值太小的实数将被归结为 0
+ 绝对值太大的实数将无法表示

```python
0.1 + 0.2
```

```python
0.2 * 0.2
```

```python
10**-500
```

```python
2.0**1024
```

<p></p>

**<font color=#0080FF>NOTE</font>：不同类型的对象混合计算时，Python 系统 (自动地) 转换到统一的类型.**

```python
3 - 2 * 2.4
```

```python
13 / 5 * 4
```

<p></p>

**<font color=#0080FF>类型转换可能出错</font>**  
+ 整数可以任意的大，可能超出浮点数类型能表示的范围
+ 复数不能转换成整数、浮点数    

```python
float(2 ** 1024)
```

**<font color=#0080FF>表达式续行规则一：遇换行符时存在明显的未完成结构 (例如，缺右括号等)</font>**

```python
(3 + 5) * (2 -
          6)
```

```python
(3 + 5) * 2 /
    5
```

<p></p>

**<font color=#0080FF>表达式续行规则二：一行末尾，即在换行符之前，是续行符 ———— </font><font color=yellow>反斜线符号 \ </font>**

```python
(3 + 5) * 2 /\
    4
```
