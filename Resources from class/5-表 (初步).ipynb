***<font color=#FF0000>Examples：表 (list) 简介及其基本操作</font>***

**<font color=#0080FF>表列举式 (display) 建立表对象</font>**  
- 语法：<font color=yellow>[表达式, ...]</font>
- 表元素可以写任意表达式，建立表时以表达式的值作为表元素
- 没有元素的表称为空表，用 [ ] 表示

```python
[1, 2, 3]            ## 三元整数表
```

```python
[1, 'math', 90.5]    ## 表中各元素类型的可以不同 
```

```python
x, y, z = 1, 2, 3
[x + 1, y**2 - 2, z]  ## 表中元素可以写任意表达式
```

<p></p>

**<font color=#0080FF>表元素的取值和赋值</font>**
- 表元素用下标表达式描述
- 合法的元素下标：0 ~ len(lst)-1 或 -len(lst) ~ -1
- <font color=yellow>表对象是可变对象，其中的元素以及表的结构可以被修改</font> (c.f. 字符串对象)

```python
lst = [1, 'math', 90.5]  ## 一个表对象是一个整体，可作为变量的值
print(lst[-1])           ## 用下标表达式取表中末尾元素
```

```python
lst[2] = 'chem' ## 也可给修改表元素的值 (表元素的赋值)
print(lst)      ## 一个表对象是一个整体，可作为函数调用时的实参
```

```python
from math import sin, cos, tan
print([sin(1.0), cos(1.0), tan(1.0)])
```

<p></p>

**<font color=#0080FF>描述下标的表达式中可以包含变量</font>**

```python
i, lst = 0, [0, 1]
i, lst[i] = 1, 2  ## 注意此处平行赋值 
print(lst)
```
